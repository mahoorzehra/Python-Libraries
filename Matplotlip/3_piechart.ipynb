```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
subject =['Math', 'English', 'SST', 'Science']
marks = [90,75,98,85]
plt.pie(marks, labels=subject)
plt.show()
```

```python
plt.pie(marks, labels=subject,
        autopct='%1.3f',
        explode=(0.1,0,0.2,0),
        colors=['purple', 'pink', "#77e2a6", "#87c6f0"], 
        wedgeprops={'edgecolor': 'black', 'linewidth': 3},  
        textprops={'fontsize': 'x-large'}, 
        shadow=True, 
        radius=1.5, 
        pctdistance=0.4, 
        labeldistance=1.5)
```
