```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
d1 = pd.read_csv(r"C:\Users\Mahoor Zehra\OneDrive\Desktop\train.csv")
```

```python
gender = d1.groupby('Sex')['Survived'].sum()
font1 = {'family':'serif','color':'purple','size':15}

plt.barh(gender.index, gender.values, color = 'purple')
plt.xlabel('No of survivers')
plt.ylabel('Gender')
plt.title('Total survived', fontdict = font1)
plt.show()
```

Stockplot:

```python
df ={
    'Year' : [2000, 2010, 2020],
    'Asia' : [3700, 4100, 4700],
    'Africa' : [800, 1050, 1350],
    'Europe' : [730, 740, 750]
}
d = pd.DataFrame(df)
d
```

```python
plt.stackplot(
    d['Year'],
    d['Asia'],
    d['Africa'],
    d['Europe'],
    labels = ['Asia', 'Africa', 'Europe'],
    colors=["#C82587", "#eee3e3", "#63038F"]
)
plt.legend()
plt.show()
```
