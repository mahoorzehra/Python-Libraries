```python
import numpy as np
```

```python
arr = np.array([21, 34, 56, 77, 91])
print(arr.ndim)                         #gives 1 i.e. one dimension
```

```python
ar1 = np.array([[21, 34, 1], [11, 32, 2]])
print(ar1)
print(ar1. ndim)
```

```python
ar2 = np.array([[21, 34, 1], [11, 32, 2], [10, 4, 21]])
print(ar2)
print(ar2.ndim)
print(ar2.shape)
print(ar2.size)
```

Hence, 3 in 1 gives a table with 3 rows and x columns (x based on the number of items in each array)
also, we can't put an irregular number of items. if we create an array of 4 columns then in every array, the number of items *must* be 4

```python
# 3D:
ar3 = np.array([
    [[1, 2, 3], [4, 5, 6]],                                                 # one table
    [[7, 8, 9], [10, 11, 12]]                                               # another table on top of it
])
print(ar3.ndim)
print(ar3.shape)
print(ar3.size)

print(ar3[0])                                                               # to access a table
print(ar3[0][1])                                                            # to access a row in the table
print(ar3[0][1][1])                                                         # to access an item in a row in the table
```
