Group by :

```python
d1['Survived'].value_counts()           # how many times 0 and 1 occured
```

```python
d1.groupby('Sex')['Survived'].sum()
```

```python
d1.groupby('Sex')['Survived'].mean()         #  mean() calculates the average survival value for each sex group.  
```

233 and 109 are representing the survivors because through sum we added all the 1s i.e. survivors. We can count deaths by reversing the encoding

```python
d1['Survived'] = d1['Survived'].map({0:1, 1:0})
```

```python
d1.groupby('Sex')['Survived'].sum()
```

```python
d1.groupby('Sex')['Survived'].mean()               #after reversing the encoding, the mean would represent the death rate.      
```

We want to count how many male and female passengers belonged to each passenger class.

```python
d1.groupby('Pclass')['Sex'].sum()                       # textual data can't be added: summing text concatenates strings
```

```python
d1['Sex'] = d1['Sex'].map({"female":0,"male":1})
```

```python
d1.groupby('Pclass')['Sex'].sum()               #gives male count
```
