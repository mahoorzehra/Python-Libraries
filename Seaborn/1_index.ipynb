```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
d = sns.load_dataset('titanic')
```

```python
d.head()
```

```python
d.shape
```

```python
grades ={
    'Subject' : ['English', 'English', 'Math', 'Math', 'SST', 'SST'],
    'Gender' : ['Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Marks' : [78, 98, 80, 98, 60, 65]
}
d1 = pd.DataFrame(grades)
d1
```

```python
sns.barplot(x = 'Subject',y='Marks',hue='Gender', data=d1)
plt.xlabel("Subjects")
```

```python
d.sample()
```

```python
sns.stripplot(x='sex', y='age', hue='pclass', data=d, palette='rocket', dodge=True, alpha=0.6)
```

```python
sns.swarmplot(x='sex', y='age', hue='pclass', data=d)
```
